<mark>有介质情形下麦克斯韦方程组</mark>：

$$
\begin{align}
\boxed{\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}} \tag{1} \\[10pt]
\boxed{\nabla \times \mathbf{H} = \mathbf{J} + \frac{\partial \mathbf{D}}{\partial t}} \tag{2} \\[10pt]
\boxed{\nabla \cdot \mathbf{D} = \rho} \tag{3} \\[10pt]
\boxed{\nabla \cdot \mathbf{B} = 0} \tag{4}
\end{align}$$

观察到相对于无介质情形下的麦克斯韦方程组有所变化，下面我们介绍上述方程组的引入过程，并且在过程中学习一些规律。

---

###1.

$$\boxed{\nabla \cdot \mathbf{D} = \rho} $$

引出该式需要学习<mark>介质的极化</mark>：

　　存在两类电介质，一类介质分子的正电中心和负电中心重合，没有电偶极矩。另一类介质分子的正负电中心不重合，有分子电偶极矩，在外场作用下，前一类的分子的正负电中心被拉开，后一类介质的分子电偶极矩平均有一定取向性，因此都出现宏观电偶极矩分布。

　　宏观电偶极矩分布用电极化强度矢量 $\mathbf{P}$ 描述，：

$$
\mathbf{P} = \frac{\sum_i \mathbf{p}_i}{\Delta V} \tag{5}
$$

　　式中 $\mathbf{p}_i$ 为第 $i$ 个分子的电偶极矩，求和符号表示对 $\Delta V$ 内所有分子求和。

　　束缚电荷密度 $\rho_P$ 和电极化强度 $\mathbf{P}$ 之间的关系：

$$
\int_V \rho_P \mathrm{d}V = -\oint_S \mathbf{P} \cdot \mathrm{d}\mathbf{S}
$$

　　上式的微分形式：

$$
\boxed{\rho_P = -\nabla \cdot \mathbf{P} }\tag{6}
$$

　　两个介质分界面薄膜上有

$$
\sigma_p \mathrm{d}S = -(\mathbf{P}_2 - \mathbf{P}_1) \cdot \mathrm{d}\mathbf{S}
$$

　　由此，

$$
\boxed{\sigma_p = - \mathbf{e}_n \cdot (\mathbf{P}_2 - \mathbf{P}_1) }\tag{7}
$$

　　$\mathbf{e}_n$ 为分界面上由介质 1 指向介质 2 的法向单位矢量。

现在我们可以完善总电荷密度$\rho$ 使其包括自由电荷密度 $\rho_f$ 和束缚电荷密度 $\rho_p$ 在内，则改写麦克斯韦方程组为：

$$
\varepsilon_0 \nabla \cdot \mathbf{E} = \rho_f + \rho_p
$$


引入电位移矢量 $\mathbf{D}$，定义为

$$
\boxed{\mathbf{D} = \varepsilon_0 \mathbf{E} + \mathbf{P} }\tag{8}
$$

前式便可写为

$$
\nabla \cdot \mathbf{D} = \rho_f
$$

实验指出，对于一般各向同性线性介质，极化强度 $\mathbf{P}$ 和 $\mathbf{E}$ 之间有简单的线性关系：

$$
\mathbf{P} = \chi_e \varepsilon_0 \mathbf{E} \tag{9}
$$

$\chi_e$ 称为介质的极化率。

$$
\mathbf{D} = \varepsilon \mathbf{E} \tag{10}
$$

$$
\varepsilon = \varepsilon_r \varepsilon_0, \quad \varepsilon_r = 1 + \chi_e \tag{11}
$$

$\varepsilon$ 称为介质的电容率，$\varepsilon_r$ 为相对电容率。

###2.
$$\boxed{\nabla \times \mathbf{H} = \mathbf{J} + \frac{\partial \mathbf{D}}{\partial t}}$$

引出该式需要学习<mark>介质的磁化</mark>：

　　介质分子内的电子运动构成微观分子电流，由于分子电流取向的无规性，没有外场时一般不出现宏观电流分布。在外磁场作用下，分子电流出现有规则取向，形成宏观磁化电流密度 $\mathbf{J}_M$。

　　分子电流可以用磁偶极矩描述。若把分子电流看作载有电流 $i$ 的小线圈，线圈面积矢量为 $\mathbf{a}$，则与分子电流相应的磁矩可表示为

$$
\mathbf{m} = i \mathbf{a}
$$

　　介质磁化后，出现宏观磁偶极矩分布，用磁化强度 $\mathbf{M}$ 表示，它定义为物理小体积 $\Delta V$ 内的总磁偶极矩与 $\Delta V$ 之比：

$$
\mathbf{M} = \frac{\sum_i \mathbf{m}_i}{\Delta V} \tag{12}
$$

　　以 $\mathbf{J}_M$ 表示磁化电流密度，有

$$
\int_S \mathbf{J}_M \cdot \mathrm{d}\mathbf{S} = \oint_L \mathbf{M} \cdot \mathrm{d}\mathbf{l}
$$

　　微分形式：

$$
\boxed{\mathbf{J}_M = \nabla \times \mathbf{M}} \tag{13}
$$

　　除了磁化电流之外，当电场变化时，介质的极化强度 $\mathbf{P}$ 发生变化，这种变化产生极化电流。


$$
\boxed{\frac{\partial \mathbf{P}}{\partial t} = \mathbf{J}_P} \tag{14}
$$

　　$\mathbf{J}_P$ 称为极化电流密度。磁化电流 $\mathbf{J}_M$ 和极化电流 $\mathbf{J}_P$ 之和是介质内的总诱导电流密度。

现在我们可以完善总电流密度$\mathbf{J}$ 使其包括自由电流$\mathbf{J}_f$，磁化电流 $\mathbf{J}_M$ ，极化电流 $\mathbf{J}_P$，位移电流密度$\mathbf{J}_D$ 在内，则改写麦克斯韦方程组为：

$$
\frac{1}{\mu_0} \nabla \times \mathbf{B} = \mathbf{J}_f + \mathbf{J}_M + \mathbf{J}_P + \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}
$$

引入磁场强度 $\mathbf{H}$，定义为

$$
\boxed{\mathbf{H} = \frac{\mathbf{B}}{\mu_0} - \mathbf{M}}\tag{15}
$$

则前式写为

$$
\nabla \times \mathbf{H} = \mathbf{J}_f + \frac{\partial \mathbf{D}}{\partial t}
$$

实验指出，对于各向同性非铁磁物质，磁化强度$\mathbf{M}$ 和 $\mathbf{H}$ 之间有简单的线性关系：

$$
\mathbf{M} = \chi_M \mathbf{H}\tag{16}
$$

$\chi_M$ 称为磁化率。

$$
\mathbf{B} = \mu \mathbf{H}\tag{17}
$$

$$
\mu = \mu_r \mu_0, \quad \mu_r = 1 + \chi_M\tag{18}
$$

$\mu$ 称为磁导率，$\mu_r$ 为相对磁导率。

补充：除了上述转换关系，在导电物质中还有欧姆定律

$$
\mathbf{J} = \sigma \mathbf{E}\tag{19}
$$
