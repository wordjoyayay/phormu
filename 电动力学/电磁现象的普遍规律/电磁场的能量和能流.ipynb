# 1.6 电磁场的能量和能流（Energy and Energy Flow of EM Field）

---

## 一、电磁场能量守恒的一般形式

电磁场是一种物质，具有能量。场能量可以在空间分布和传播，也能与电荷系统相互转化。描述电磁场能量需要两个物理量：

- **能量密度** $w(\boldsymbol{x}, t)$：单位体积内的电磁能量
- **能流密度矢量** $\boldsymbol{S}$（坡印廷矢量）：单位时间通过单位面积的能量，方向指向能量传播方向

考虑空间某体积 $V$，界面为 $S$。能量守恒要求：

> **单位时间通过界面流入 $V$ 的能量 = $V$ 内电磁能量的增加率 + 场对电荷系统做功的功率**

数学表达为：

$$
-\oint_S \boldsymbol{S} \cdot d\boldsymbol{\sigma} = \int_V \boldsymbol{f} \cdot \boldsymbol{v}\, dV + \frac{d}{dt}\int_V w\, dV
\qquad\text{(1)}
$$

对应的微分形式：

$$
\boxed{\nabla \cdot \boldsymbol{S} + \frac{\partial w}{\partial t} = -\boldsymbol{J} \cdot \boldsymbol{E}}
\qquad\text{(2)}
$$

> 式(2)是电磁场能量守恒的**核心方程**，左边描述能量在空间中的流动和变化，右边是场对电荷系统做功的功率密度。

若 $V$ 为全空间，则面积分为零，得到：

$$
\int_V \boldsymbol{f} \cdot \boldsymbol{v}\, dV = -\frac{d}{dt}\int_V w\, dV
\qquad\text{(3)}
$$

> 场对电荷做功的总功率等于电磁场总能量的减少率——场和电荷的总能量守恒。

---

## 二、能流密度和能量密度的具体形式

利用麦克斯韦方程组可从式(2)反推出 $\boldsymbol{S}$ 和 $w$ 的表达式。

由 $\boldsymbol{J} = \nabla \times \boldsymbol{H} - \dfrac{\partial \boldsymbol{D}}{\partial t}$（麦克斯韦第二式），有：

$$
\boldsymbol{J} \cdot \boldsymbol{E} = \boldsymbol{E} \cdot (\nabla \times \boldsymbol{H}) - \boldsymbol{E} \cdot \frac{\partial \boldsymbol{D}}{\partial t}
$$

利用矢量恒等式 $\nabla \cdot (\boldsymbol{E} \times \boldsymbol{H}) = \boldsymbol{H} \cdot (\nabla \times \boldsymbol{E}) - \boldsymbol{E} \cdot (\nabla \times \boldsymbol{H})$ 及 $\nabla \times \boldsymbol{E} = -\dfrac{\partial \boldsymbol{B}}{\partial t}$，得到：

$$
\boldsymbol{J} \cdot \boldsymbol{E} = -\nabla \cdot (\boldsymbol{E} \times \boldsymbol{H}) - \boldsymbol{H} \cdot \frac{\partial \boldsymbol{B}}{\partial t} - \boldsymbol{E} \cdot \frac{\partial \boldsymbol{D}}{\partial t}
$$

与式(2)比较，可得：

**能流密度矢量（坡印廷矢量）：**

$$
\boxed{\boldsymbol{S} = \boldsymbol{E} \times \boldsymbol{H}}
\qquad\text{(4)}
$$

**能量密度变化率：**

$$
\boxed{\frac{\partial w}{\partial t} = \boldsymbol{E} \cdot \frac{\partial \boldsymbol{D}}{\partial t} + \boldsymbol{H} \cdot \frac{\partial \boldsymbol{B}}{\partial t}}
\qquad\text{(5)}
$$

## 三、线性介质的能量密度

对于线性介质，本构关系为 $\boldsymbol{D} = \varepsilon \boldsymbol{E}$，$\boldsymbol{B} = \mu \boldsymbol{H}$，式(5)可以积分得到：

$$
\boxed{w = \frac{1}{2}(\boldsymbol{E} \cdot \boldsymbol{D} + \boldsymbol{H} \cdot \boldsymbol{B})}
\qquad\text{(6)}
$$

> 注意：式(6)仅适用于线性介质。更一般的公式是式(5)，它对**任何介质**（包括非线性介质）都成立。

**真空中**（$\varepsilon = \varepsilon_0,\ \mu = \mu_0$）：

$$
\boldsymbol{S} = \frac{1}{\mu_0} \boldsymbol{E} \times \boldsymbol{B}, \qquad
w = \frac{1}{2}\!\left(\varepsilon_0 E^2 + \frac{1}{\mu_0} B^2\right)
\qquad\text{(7)}
$$

## 四、能量在电路中的传播——同轴传输线示例

直流电路的直觉认为能量是通过导线内的电子输运的，但这一节会展示一个更深刻的图像：**能量实际上是通过电磁场在空间传播的**。

### 问题

同轴线内导体半径 $a$，外导体半径 $b$，其间为绝缘介质。内导体载流 $I$，内外导体间电压 $U$。求：
1. 忽略导线电阻时，能流密度 $\boldsymbol{S}$ 和传输功率
2. 计入导线电导率 $\sigma$，证明能量经表面进入导线内部，提供欧姆损耗

### 解

**(1) 理想导体情况**

由对称性，在内导体外（$a<r<b$），安培环路定理和 Gauss 定理给出：

$$
\boldsymbol{H} = \frac{I}{2\pi r}\,\boldsymbol{e}_\phi, \qquad
\boldsymbol{E} = \frac{\tau}{2\pi\varepsilon r}\,\boldsymbol{e}_r
$$

其中 $\tau$ 是内导线单位长度电荷。内外导体间电压 $U = \int_a^b E\,dr = \dfrac{\tau}{2\pi\varepsilon}\ln\!\dfrac{b}{a}$。

能流密度：

$$
\boldsymbol{S} = \boldsymbol{E} \times \boldsymbol{H} = \frac{I\tau}{4\pi^2\varepsilon r^2}\,\boldsymbol{e}_z
$$

传输功率：

$$
P = \int_S \boldsymbol{S} \cdot d\boldsymbol{\sigma} = \int_a^b S \cdot 2\pi r\,dr = UI
\qquad\text{(8)}
$$

> 这正是电路理论中的功率表达式 $P = UI$，但这里它来自**场在空间中的传播**，而非导线内部的传导。

**(2) 实际导体——欧姆损耗**

计入导线的有限电导率 $\sigma$，由欧姆定律 $\boldsymbol{J} = \sigma \boldsymbol{E}$，导线内部电场沿 $z$ 方向，大小为：

$$
E_z = \frac{J}{\sigma} = \frac{I}{\sigma \pi a^2}
$$

于是导线表面附近的 $\boldsymbol{S}$ 出现径向分量 $\boldsymbol{S}_\perp$（指向导线内部）：

$$
S_\perp = -\frac{I^2}{2\pi^2 a^3 \sigma}
$$

流入长度为 $\Delta l$ 导线的功率即为损耗功率：

$$
P_{\text{loss}} = -S_\perp \cdot 2\pi a\,\Delta l = I^2 R
\qquad\text{(9)}
$$

> 这说明导线电阻的**焦耳热**来源于电磁场能量从外部空间**垂直流入**导线内部，而非电流本身携带能量。

## 小结

本节的核心逻辑：

```
电磁场具有能量，可在空间传播
      ↓
能量守恒定律 → 连续性方程 ∇·S + ∂w/∂t = -J·E
      ↓
用 Maxwell 方程组反推
      ↓
S = E × H（坡印廷矢量）        ∂w/∂t = E·∂D/∂t + H·∂B/∂t
      ↓
线性介质：w = ½(E·D + H·B)
      ↓
同轴线算例 → 能量通过场在空间传输，而非导线内部
                    导线电阻的焦耳热来源于场能径向流入
```

> **关键思想**：能量不是通过导线内部传导的，而是通过导线周围的**电磁场**以坡印廷矢量的形式传播的。导线的作用只是**引导**电磁场的分布和能量走向。
