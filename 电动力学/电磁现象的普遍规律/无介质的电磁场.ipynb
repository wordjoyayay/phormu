<mark>无介质情形下麦克斯韦方程组</mark>：

$$
\begin{align}
\boxed{\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}} \tag{1} \\[10pt]
\boxed{\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}} \tag{2} \\[10pt]
\boxed{\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}} \tag{3} \\[10pt]
\boxed{\nabla \cdot \mathbf{B} = 0} \tag{4}
\end{align}
$$

该方程组为电动力学基本公理，无需证明，不过可以解释一下方程组的由来，并且我们可以在此过程中学习各种定律及概念：

---

###1.
$$ \nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0} $$

想要引出这个方程，需要了解<mark>库仑定律</mark>：

　　真空中静止点电荷 $ Q $ 对另一个静止点电荷 $ Q' $ 的作用力 $ \mathbf{F} $ 为

$$
\mathbf{F} = \frac{Q Q'}{4 \pi \varepsilon_0 r^3} \mathbf{r}
$$

　　一个静止点电荷 $ Q $ 所激发的电场强度为

$$
\mathbf{E} = \frac{Q \mathbf{r}}{4 \pi \varepsilon_0 r^3}
$$

　　设由源点 $ x' $ 到场点 $ x $ 的矢径为 $\mathbf{r}$，则 $ P $ 点上的电场强度 $ \mathbf{E} $ 为

$$
\boxed{\mathbf{E}(x) = \int_{V} \frac{\rho(x') \mathbf{r}}{4 \pi \varepsilon_0 r^3} \mathrm{d}V'} \tag{5}
$$

<img src="https://github.com/wordjoyayay/pictures/raw/main/coulomb_force_vectors.png" width="300" style="display: block; margin: 0 auto;">

由库仑定律可以推出<mark>高斯定理</mark>：

$$
\oint_{S} \mathbf{E} \cdot \mathrm{d}\mathbf{S} = \frac{Q}{\varepsilon_0} \tag{6}
$$

证明过程如下：

<img src="https://github.com/wordjoyayay/pictures/raw/main/gauss_law_proof.png" width="600">

以上为高斯定理的积分形式，由散度变换式得出微分形式：
$$
\boxed{\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}} \tag{3}
$$

该式表达了电场的散度，为麦克斯韦方程组一支。（值得注意，库仑定律仅适用于静电情形，而该式为基本公理，所以也适用于运动电荷情形）

---
































### 2.
$$\nabla \cdot \mathbf{B} = 0$$

类似于库仑定律引出电场散度，磁场的散度引出也需要一个定律，即<mark>毕奥-萨伐尔定律</mark>：

　　实验指出，一个电流元 $ I\mathrm{d}\mathbf{l} $ 在磁场中所受的力可以表为

$$
\mathrm{d}\mathbf{F} = I\mathrm{d}\mathbf{l} \times \mathbf{B}
$$

　　其中磁感应强度$\mathbf{B}$：

$$
\boxed{\mathbf{B}(\mathbf{x}) = \frac{\mu_0}{4\pi} \int_{V} \frac{\mathbf{J}(\mathbf{x}') \times \mathbf{r}}{r^3} \mathrm{d}V'} \tag{7}
$$

　　细导线上恒定电流激发磁场时写为

$$
\mathbf{B}(\mathbf{x}) = \frac{\mu_0}{4\pi} \oint_{L} \frac{I\mathrm{d}\mathbf{l} \times \mathbf{r}}{r^3}
$$

由毕奥-萨伐尔定律推导磁场散度过程如下：

<img src="https://github.com/wordjoyayay/pictures/raw/main/magnetic_divergence_proof.png" width="600">

其中算符$\nabla$为$x$的算符值得注意，以及矢量$\mathbf{A}$为磁矢势，在其他章节将仔细讲解。

$$
\boxed{\nabla \cdot \mathbf{B} = 0} \tag{4}
$$

由此得出磁场的散度表达式，也是麦克斯韦方程组一支。

### 3.
$$\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}$$

该方程首先需要库仑定律建造雏形，推导过程如下：

<img src="https://github.com/wordjoyayay/pictures/raw/main/electric_curl_proof.png" width="600">

以上为积分形式，微分形式可由旋度变换式得出：

$$
\boxed{\nabla \times \mathbf{E} = 0} \tag{8}
$$

上式表示了静电场的无旋性，然而实验发现不仅电荷可以激发电场，磁场也可以激发感应电场，因此该式还可以进行完善。

学习<mark>电磁感应定律</mark>：

$$
\boxed{\mathcal{E} = -\frac{\mathrm{d}}{\mathrm{d}t} \int_{S} \mathbf{B} \cdot \mathrm{d}\mathbf{S}} \tag{9}
$$

线圈上的电荷是直接受到该处电场作用而运动的，线圈上有感应电流就表明空间中存在着电场。因此，电磁感应现象的实质是变化磁场在其周围空间中激发了电场。

感应电动势是电场强度沿闭合回路的线积分，因此电磁感应定律 (9) 式可写为

$$
\oint_{L} \mathbf{E} \cdot \mathrm{d}\mathbf{l} = -\frac{\mathrm{d}}{\mathrm{d}t} \int_{S} \mathbf{B} \cdot \mathrm{d}\mathbf{S}
$$

若回路 $ L $ 是空间中的一条固定回路，则上式为

$$
\oint_{L} \mathbf{E} \cdot \mathrm{d}\mathbf{l} = -\int_{S} \frac{\partial \mathbf{B}}{\partial t} \cdot \mathrm{d}\mathbf{S}
$$

化为微分形式后得

$$
\boxed{\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}} \tag{1}
$$

这是磁场对电场作用的基本规律。由该式可见，感应电场是有旋场。在一般情况下，总电场包括静电场和感应电场，因此应由 $(1)$ 式表示一般电场的旋度。

### 4.
$$\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}$$

该方程首先需要毕奥-萨伐尔定律建造雏形，推导过程如下：

<img src="https://github.com/wordjoyayay/pictures/raw/main/magnetic_curl_proof.png" width="600">




















该式还可由<mark>安培环路定律</mark>得出，定律如下：

$$
\boxed{\oint_{L} \mathbf{B} \cdot \mathrm{d}\mathbf{l} = \mu_0 I} \tag{10}
$$

不过因为安培环路定律也可以由毕奥-萨伐尔定律推得，此处不再赘述。

由此可得：

$$
\boxed{\nabla \times \mathbf{B} = \mu_0 \mathbf{J}} \tag{11}
$$

该式适用于恒定电流情形，若是非恒定电流则不适用，下面我们来说明这点。首先需要了解<mark>电荷守恒定律</mark>：

　　如果有电荷从该区域流出的话，区域 % V $ 内的电荷必然减小。通过界面流出的总电流应该等于$ V $ 内的电荷减小率，即

$$
\oint_{S} \mathbf{J} \cdot \mathrm{d}\mathbf{S} = -\int_{V} \frac{\partial \rho}{\partial t} \mathrm{d}V
$$

　　微分形式

$$
\boxed{\nabla \cdot \mathbf{J} + \frac{\partial \rho}{\partial t} = 0} \tag{12}
$$

　　上式称为<mark>电流连续性方程</mark>，它是电荷守恒定律的微分形式。

　　以上公式是对任意变化电流成立的。在恒定电流情况下，一切物理量不随时间而变，因而 $ \partial \rho / \partial t = 0 $，因此由 (12) 式得

$$
\nabla \cdot \mathbf{J} = 0 \quad (\text{恒定电流}) \tag{13}
$$

　　上式表示恒定电流的连续性。恒定电流分布是无源的，其流线必为闭合曲线，没有发源点和终止点。换句话说，恒定电流（直流电）只能够在闭合回路中通过，电路一断，直流电就不能通过，这是我们熟知的事实。

**非恒定电流情况下：**

$$
\nabla \cdot \mathbf{J} = -\frac{\partial \rho}{\partial t} \neq 0
$$

如果我们之前得出的 (11) 式两边同取散度：

$$
\nabla \cdot (\nabla \times \mathbf{B}) = 0 = \mu_0 \nabla \cdot \mathbf{J}
$$

则与非恒定电流情况下的电流连续性方程出现矛盾，但电流连续性方程是由电荷守恒定律推出，是基础方程，因此我们选择更改之前的 (11) 式为：

$$
\nabla \cdot (\mathbf{J} + \mathbf{J}_D) = 0 \tag{14}
$$

其中 $ \mathbf{J}_D $ 称为位移电流。

由电流连续性方程和电场散度表达式：

$$
\nabla \cdot \mathbf{J} + \frac{\partial \rho}{\partial t} = 0
$$

$$
\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}
$$

两式合起来得

$$
\nabla \cdot \left( \mathbf{J} + \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t} \right) = 0
$$

与 (14) 式比较即得 $ \mathbf{J}_D $ 的一个可能表示式（还有别的可能的表达式，不过先取简单的一式）：

$$
\mathbf{J}_D = \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}
$$

综上，得到满足一般磁场的旋度表达式，即麦克斯韦方程组一支：

$$
\boxed{\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}} \tag{2}
$$

该式中的 $ \mathbf{J} $ 为传导电流。
