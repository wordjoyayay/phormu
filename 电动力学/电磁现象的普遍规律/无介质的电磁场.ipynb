<mark>无介质情形下麦克斯韦方程组</mark>：
$$
\begin{align}
\boxed{\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}} \tag{1} \\[10pt]
\boxed{\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}} \tag{2} \\[10pt]
\boxed{\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}} \tag{3} \\[10pt]
\boxed{\nabla \cdot \mathbf{B} = 0} \tag{4}
\end{align}
$$
该方程组为电动力学基本公理，无需证明，不过可以解释一下方程组的由来，并且我们可以在此过程中学习各种定律及概念：

###1.
$$\nabla \cdot \mathbf{B} = 0$$

想要引出这个方程，需要了解<mark>库仑定律</mark>：

　　真空中静止点电荷 $ Q $ 对另一个静止点电荷 $ Q' $ 的作用力 $ \mathbf{F} $ 为

$$
\mathbf{F} = \frac{Q Q'}{4 \pi \varepsilon_0 r^3} \mathbf{r}
$$



　　一个静止点电荷 $ Q $ 所激发的电场强度为

$$
\mathbf{E} = \frac{Q \mathbf{r}}{4 \pi \varepsilon_0 r^3} \tag{5}
$$

设由源点 $ x' $ 到场点 $ x $ 的矢径为 $\mathbf{r}$，则 $ P $ 点上的电场强度 $ \mathbf{E} $ 为

$$
\mathbf{E}(x) = \int_{V} \frac{\rho(x') \mathbf{r}}{4 \pi \varepsilon_0 r^3} \mathrm{d}V' \tag{1.5}
$$
