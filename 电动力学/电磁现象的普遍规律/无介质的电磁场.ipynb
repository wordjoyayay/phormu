无介质情形下麦克斯韦方程组：
\begin{align}
\nabla \times \mathbf{E} &= -\frac{\partial \mathbf{B}}{\partial t} \tag{1} \\[10pt]
\nabla \times \mathbf{B} &= \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t} \tag{2} \\[10pt]
\nabla \cdot \mathbf{E} &= \frac{\rho}{\varepsilon_0} \tag{3} \\[10pt]
\nabla \cdot \mathbf{B} &= 0 \tag{4}
\end{align}
