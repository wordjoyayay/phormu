电动力学所需矢量分析公式：

散度变换式：
$$
\int_{V} (\nabla \cdot \mathbf{f}) \, \mathrm{d}V=\oint_{S} \mathbf{f} \cdot \mathrm{d}\mathbf{S}
$$
旋度变换式：
$$
\int_{S} (\nabla \times \mathbf{f}) \cdot \mathrm{d}\mathbf{S}=\oint_{L} \mathbf{f} \cdot \mathrm{d}\mathbf{l}
$$



标量场的梯度必为无旋场：
$$
\nabla \times (\nabla \varphi) = 0
$$


矢量场的旋度必为无源场：
$$
\nabla \cdot (\nabla \times \mathbf{f}) = 0
$$





三矢量矢积：

$$
(\mathbf{a} \times \mathbf{b}) \times \mathbf{c} = (\mathbf{c} \cdot \mathbf{a})\mathbf{b} - (\mathbf{c} \cdot \mathbf{b})\mathbf{a}
$$







$\nabla$算符运算公式

公式中 $\varphi,\psi$ 代表标量场，$\mathbf{f},\mathbf{g}$ 代表矢量场。
$$
\nabla(\varphi\psi) = \varphi\nabla\psi + \psi\nabla\varphi
$$

$$
\nabla\cdot(\varphi\mathbf{f}) = (\nabla\varphi)\cdot\mathbf{f} + \varphi\nabla\cdot\mathbf{f}
$$

$$
\nabla\times(\varphi\mathbf{f}) = (\nabla\varphi)\times\mathbf{f} + \varphi\nabla\times\mathbf{f}
$$

$$
\nabla\cdot(\mathbf{f}\times\mathbf{g}) = (\nabla\times\mathbf{f})\cdot\mathbf{g} - \mathbf{f}\cdot(\nabla\times\mathbf{g})
$$

$$
\nabla\times(\mathbf{f}\times\mathbf{g}) = (\mathbf{g}\cdot\nabla)\mathbf{f} + (\nabla\cdot\mathbf{g})\mathbf{f} - (\mathbf{f}\cdot\nabla)\mathbf{g} - (\nabla\cdot\mathbf{f})\mathbf{g}
$$

$$
\nabla(\mathbf{f}\cdot\mathbf{g}) = \mathbf{f}\times(\nabla\times\mathbf{g}) + (\mathbf{f}\cdot\nabla)\mathbf{g} + \mathbf{g}\times(\nabla\times\mathbf{f}) + (\mathbf{g}\cdot\nabla)\mathbf{f}
$$

$$
\nabla\cdot\nabla\varphi \equiv \nabla^2\varphi
$$

$$
\nabla\times(\nabla\times\mathbf{f}) = \nabla(\nabla\cdot\mathbf{f}) - \nabla^2\mathbf{f}
$$
